### Exercise 8 — An artist profile
1. `intro` — `"Coldplay (GB) plays rock"`, built with an f-string from the dict
2. Coldplay gained 800 monthly listeners: increase `"monthly_listeners"` by 800
3. Add the key `"albums"` with the value `9`
4. `label` — the record label. It is **not** in the dict: use `.get` with the default `"unknown"`

````python
artist = {"name": "Coldplay", "country": "GB", "genre": "rock", "monthly_listeners": 41200}

intro = f"{artist['name']} ({artist['country']}) plays {artist['genre']}"

artist["monthly_listeners"] += 800

artist["albums"] = 9

label = artist.get("label", "unknown")

print(intro)  # expected: Coldplay (GB) plays rock
print(artist["monthly_listeners"], artist["albums"], label)  # expected: 42000 9 unknown
````

> **Careful with quotes:** inside `f"..."`, write the keys with single quotes: `f"{artist['name']}"`.
> Using `"` inside `"..."` works only in Python 3.12 and newer.

### Exercise 9 (Stretch) — A family-friendly playlist
`tracks` is a small table: a list of dicts. Using a loop:
1. `explicit_count` — how many tracks are explicit
2. `family_minutes` — total length in minutes of the tracks that are **not** explicit, rounded to 1 decimal
3. Print the title of every track shorter than 3 minutes

````python
tracks = [
    {"title": "Neon Rain", "artist": "Dua Lipa", "seconds": 203, "explicit": False},
    {"title": "Midnight City Lights", "artist": "Drake", "seconds": 245, "explicit": True},
    {"title": "Paper Planes", "artist": "Coldplay", "seconds": 268, "explicit": False},
    {"title": "Golden Hour", "artist": "Billie Eilish", "seconds": 176, "explicit": False},
    {"title": "Fast Lane", "artist": "Eminem", "seconds": 231, "explicit": True},
]

explicit_count = 0
family_seconds = 0

for track in tracks:
    if track["explicit"]:
        explicit_count+= 1
    else:
        family_seconds += track["seconds"]

    if track["seconds"] < 180:
        print(track["title"])

family_minutes = round(family_seconds / 60 , 1)

print(explicit_count, family_minutes)
# expected:
# Golden Hour
# 2 10.8
````

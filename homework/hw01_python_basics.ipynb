### Exercise 1 — Pretty durations
Turn `ms = 185000` into the text `"3:05"` and store it in `pretty`.

Careful: `f"{5}"` gives `"5"`, but `f"{5:02d}"` gives `"05"` (at least 2 digits, padded with zeros).

```python
ms = 185000

seconds = ms // 1000

minutes = seconds // 60

seconds = seconds % 60

pretty = f"{minutes}:{seconds:02d}"

# YOUR CODE HERE

print(pretty)  # expected: 3:05
```

### Exercise 4 — A small playlist
Using `playlist` below (no loops needed):
1. `total_ms` — total duration of the playlist in milliseconds
2. `total_text` — the same in `m:ss`
3. `longest` — the longest duration in ms (hint: make a list of the 4 durations, then use `max`)
4. `has_coldplay` — `True` if `"Coldplay"` is one of the artists (hint: list + `in`)

```python
playlist = [
    {"title": "Golden Hearts", "artist": "Coldplay", "duration_ms": 241000},
    {"title": "Neon Rain", "artist": "Dua Lipa", "duration_ms": 203000},
    {"title": "Silent Waves", "artist": "Hans Zimmer", "duration_ms": 412000},
    {"title": "Paper Dreams", "artist": "Ed Sheeran", "duration_ms": 187000},
]

# YOUR CODE HERE

total_ms = playlist[0]["duration_ms"] + playlist[1]["duration_ms"] +playlist[2]["duration_ms"] + playlist[3]["duration_ms"]
print(total_ms)

total_seconds = total_ms // 1000
total_minutes = total_seconds // 60
seconds = total_seconds % 60 

total_text = f"{total_minutes}:{seconds:02d}"
print(total_text)

durations = [playlist[0]["duration_ms"] , playlist[1]["duration_ms"] ,playlist[2]["duration_ms"],playlist[3]["duration_ms"]]
longest = max(durations)

artists = [playlist[0]["artist"] , playlist[1]["artist"] , playlist[2]["artist"] , playlist[3]["artist"]]
has_coldplay = "Coldplay" in artists
print(has_coldplay)


print(total_ms, total_text, longest, has_coldplay)  # expected: 1043000 17:23 412000 True
```

### Exercise 5 — A nested app event
From `event` get:
1. `user_id`
2. `device_os` (the operating system)
3. `n_tracks` — how many tracks were added to the playlist
4. `first_track` — the first track id
5. `channel` — the share channel; this event has none, so use `"none"` as default

```python
event = {
    "event_id": "50c328696212",
    "event_type": "playlist_add",
    "user": {"id": "U0788", "plan": "free", "device": {"type": "mobile", "os": "ios", "app_version": "5.3.0"}},
    "payload": {"playlist_name": "Road Trip", "track_ids": ["T0110", "T0042", "T0007"]},
}

user_id = event["user"]["id"]
print(user_id)

device_os = event["user"]["device"]["os"]
print(device_os)

n_tracks = len(event["payload"]["track_ids"])
print(n_tracks)

first_track = event["payload"]["track_ids"][0]
print(first_track)

channel = event.get("channel", "none")
print(channel)

print(user_id, device_os, n_tracks, first_track, channel)  # expected: U0788 ios 3 T0110 none
```

### Exercise 6 (Stretch) — From a text line to a row
An old system exports plays as text lines separated by `;`. Turn `line` into a dict `row` with keys
`user_id`, `artist`, `title`, `seconds` where `seconds` is an `int` (the duration is in `m:ss`).

```python
line = "U0525;  dua lipa ;Neon Rain;3:45"


parts = line.split(";")
correct_parts = [part.strip().title() for part in parts]

print(correct_parts)

list = correct_parts[3].split(":")
print(list)
minutes = int(list[0])
seconds = int(list[1])
print(minutes)
print(seconds)

total = minutes  * 60 + seconds

row = {"user_id" : correct_parts[0],
       "artist" : correct_parts[1] ,
       "title" : correct_parts[2],
       "seconds" : total 
       }


print(row)  # expected: {'user_id': 'U0525', 'artist': 'Dua Lipa', 'title': 'Neon Rain', 'seconds': 225}
```
